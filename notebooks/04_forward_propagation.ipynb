```python
import numpy as np
```

Aktivasyon Fonksiyonu (Sigmoid)

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
```

```python
X = np.array([[1.5, 0.5]])
```

```python
W1 = np.array([[0.2, 0.8, -0.5],
               [0.5, -0.1, 0.2]])
b1 = np.array([[0.1, 0.1, 0.1]])
```

```python
W2 = np.array([[0.4],
               [0.9],
               [-0.3]])
b2 = np.array([[0.2]])
```

```python
z1 = np.dot(X, W1) + b1
a1 = sigmoid(z1)
```

```python
z2 = np.dot(a1, W2) + b2
y_tahmin = sigmoid(z2)
```

```python
print("Gizli Katman Çıktısı (a1):", a1)
print("Final Tahmin (y_tahmin):", y_tahmin)
```
