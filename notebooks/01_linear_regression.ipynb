```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)

m = 20
x = np.linspace(0,10,m)
y = 2.5 * x + 3 + np.random.normal(0, 2, size=m)
```

```python
x
```

```python
y
```

```python
plt.scatter(x,y)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Training data")
plt.show()
```

```python
def f_wb(x, w, b):
    return w * x  + b
```

```python
plt.scatter(x, y)

for (w,b) in [(0,0),(1,0),(2.5,3),(3.5,1)]:
    y_hat = f_wb(x, w, b)
    plt.plot(x, y_hat, label = f"w={w}, b={b}")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Different lines with different (w,b)")
plt.show()
```
